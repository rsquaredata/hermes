# Employment

This notebook prepares the municipality-level employment dataset used by HERMES.

## Pipeline

1. Download
2. Load
3. Explore
4. Validate raw dataset
5. Prepare
6. Validate prepared dataset
7. Save

## Output

- `employment.parquet`

```python
# ============================================================================
# Imports
# ============================================================================

from hermes.config import PREPARED_DIR
from hermes.io import save_dataframe
from hermes.loaders import load_employment_raw
from hermes.preprocessing import prepare_employment
from hermes.sources import download_employment
from hermes.validation import (
    explore_dataframe,
    validate_dataframe,
)
```

```python
# ============================================================================
# Download dataset
# ============================================================================

download_employment()
```

```python
# ============================================================================
# Load dataset
# ============================================================================

employment_raw = load_employment_raw()
```

```python
# ============================================================================
# Explore raw dataset
# ============================================================================

explore_dataframe(
    employment_raw,
    title="Raw Employment Dataset",
)
```

```python
# ============================================================================
# Validate raw dataset
# ============================================================================

validate_dataframe(
    employment_raw,
    expect_missing=True,
)
```

```python
# ============================================================================
# Prepare dataset
# ============================================================================

employment = prepare_employment(
    employment_raw,
)

employment.head()
```

```python
# ============================================================================
# Explore prepared dataset
# ============================================================================

explore_dataframe(
    employment,
    title="Prepared Employment Dataset",
)
```

```python
# ============================================================================
# Validate prepared dataset
# ============================================================================

validate_dataframe(
    employment,
    key="insee_code",
    expect_missing=True
)
```

```python
# ============================================================================
# Save dataset
# ============================================================================

save_dataframe(
    employment,
    PREPARED_DIR / "employment.parquet",
)
```
