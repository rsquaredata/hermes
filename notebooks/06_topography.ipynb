# Topography

This notebook prepares the municipality-level topography dataset used by HERMES.

## Pipeline

1. Download dataset
2. Load dataset
3. Explore raw dataset
4. Validate raw dataset
5. Prepare dataset
6. Explore prepared dataset
7. Validate prepared dataset
8. Save dataset

## Output

- `topography.parquet`

```python
# ============================================================================
# Imports
# ============================================================================

from hermes.config import PREPARED_DIR
from hermes.io import save_dataframe
from hermes.loaders import load_topography_raw
from hermes.preprocessing import prepare_topography
from hermes.sources import download_topography
from hermes.validation import (
    explore_dataframe,
    validate_dataframe,
)
```

```python
# ============================================================================
# Download dataset
# ============================================================================

download_topography()
```

```python
# ============================================================================
# Load dataset
# ============================================================================

topography_raw = load_topography_raw()
```

```python
# ============================================================================
# Explore raw dataset
# ============================================================================

explore_dataframe(
    topography_raw,
    title="Raw Topography Dataset",
)
```

```python
(
    topography_raw
    .isna()
    .sum()
    .loc[lambda s: s > 0]
    .sort_values(ascending=False)
)
```

```python
# ============================================================================
# Validate raw dataset
# ============================================================================

validate_dataframe(
    topography_raw,
)
```

```python
# ============================================================================
# Prepare dataset
# ============================================================================

topography = prepare_topography(
    topography_raw,
)
```

```python
# ============================================================================
# Explore prepared dataset
# ============================================================================

explore_dataframe(
    topography,
    title="Prepared Topography Dataset",
)
```

```python
(
    topography
    .isna()
    .sum()
    .loc[lambda s: s > 0]
    .sort_values(ascending=False)
)
```

```python
topography.columns.tolist()
```

```python
topography[
    topography["mean_elevation"].isna()
][
    [
        "insee_code",
        "area_hectares",
        "mean_elevation",
        "latitude",
        "longitude",
    ]
]
```

```python
from hermes.loaders import load_municipality_boundaries

municipality_boundaries = load_municipality_boundaries()

municipality_boundaries[
    municipality_boundaries["insee_code"].isin(
        topography.loc[
            topography["mean_elevation"].isna(),
            "insee_code",
        ]
    )
][
    [
        "insee_code",
        "municipality",
        "geometry",
    ]
]
```

```python
topography.isna().sum().sort_values(ascending=False)
```

```python
topography.loc[
    topography["area_hectares"].isna(),
    ["insee_code"]
]
```

```python
topography.loc[
    topography["mean_elevation"].isna(),
    :
]
```

```python
municipality_boundaries.loc[
    municipality_boundaries["insee_code"].isin(
        [
            "12218",
            "14581",
            "15171",
            "49126",
            "69114",
        ]
    ),
    ["insee_code", "municipality"]
]
```

<span style="color:green">
<b>Expected missing values:</b>

The prepared topography dataset contains a limited number of expected missing values inherited from the official source data.

- `area_hectares` is missing for 129 overseas municipalities.
- Elevation and centroid coordinates are missing for five recently created French municipalities ("communes nouvelles"):
  - Conques-en-Rouergue (12218)
  - Aurseulles (14581)
  - Sainte-Anastasie (15171)
  - Orée d'Anjou (49126)
  - Porte des Pierres Dorées (69114)

These missing values originate from the source datasets and do not indicate preprocessing errors.
</span>

```python
# ============================================================================
# Validate prepared dataset
# ============================================================================

validate_dataframe(
    topography,
    key="insee_code",
    expect_missing=True,
)
```

```python
# ============================================================================
# Save dataset
# ============================================================================

save_dataframe(
    topography,
    PREPARED_DIR / "topography.parquet",
)
```
