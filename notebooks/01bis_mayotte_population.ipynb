```python
# ============================================================================
# Imports
# ============================================================================

import pandas as pd

from hermes.data_catalog import get_dataset
from hermes.io import download_file
from hermes.preprocessing import prepare_mayotte_population
```

```python
# ============================================================================
# Get dataset
# ============================================================================

dataset = get_dataset(
    "mayotte_population_raw"
)

dataset
```

```python
# ============================================================================
# Download dataset
# ============================================================================

download_file(
    url=dataset.url,
    destination=dataset.download_path,
)
```

```python
# ============================================================================
# Inspect workbook
# ============================================================================

workbook = pd.ExcelFile(
    dataset.local_path
)

workbook.sheet_names
```

```python
# Inspect workbook summary

summary = pd.read_excel(
    dataset.local_path,
    sheet_name="Sommaire",
    engine="xlrd",
    header=None,
)

summary.head(30)
```

```python
# Inspect IND1

ind1_raw = pd.read_excel(
    dataset.local_path,
    sheet_name="IND1",
    engine="xlrd",
    header=None,
)

ind1_raw.head(30)
```

```python
# Inspect IND8

ind8_raw = pd.read_excel(
    dataset.local_path,
    sheet_name="IND8",
    engine="xlrd",
    header=None,
)

ind8_raw.head(30)
```

```python
# ============================================================================
# Load Mayotte population
# ============================================================================

mayotte_population_raw = pd.read_excel(
    dataset.local_path,
    sheet_name="IND1",
    engine="xlrd",
    header=None,
)
```

```python
# ============================================================================
# Prepare Mayotte population
# ============================================================================

mayotte_population = prepare_mayotte_population(
    mayotte_population_raw,
)

mayotte_population
```

```python
from hermes.integration import integrate_population
from hermes.loaders import load_population
from hermes.config import PREPARED_DIR
from hermes.io import save_dataframe

# Call existing prepared population dataset missing Mayotte data
population = load_population()
```

```python
# Integrate missing Mayotte population data to prepared population dataset
population_integrated = integrate_population(
    population,
    mayotte_population,
)

# Check integration
print(population.shape)
print(mayotte_population.shape)
print(population_integrated.shape)
```

```python
# ============================================================================
# Save updated population dataset
# ============================================================================

save_dataframe(
    population_integrated,
    PREPARED_DIR / "population.parquet",
)
```
