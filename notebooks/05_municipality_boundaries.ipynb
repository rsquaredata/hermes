# Municipality Boundaries

This notebook prepares the municipality boundaries dataset used by HERMES.

## Pipeline

1. Download dataset
2. Load dataset
3. Explore raw dataset
4. Validate raw dataset
5. Prepare dataset
6. Explore prepared dataset
7. Validate prepared dataset
8. Save dataset

## Output

- `municipality_boundaries.parquet`

```python
# ============================================================================
# Imports
# ============================================================================

import geopandas as gpd

from hermes.config import PREPARED_DIR
from hermes.data_catalog import get_dataset
from hermes.io import save_dataframe
from hermes.loaders import load_municipality_boundaries
from hermes.preprocessing import prepare_municipality_boundaries
from hermes.sources import download_municipality_boundaries
from hermes.validation import (
    explore_dataframe,
    validate_dataframe,
)
```

```python
# ============================================================================
# Download dataset
# ============================================================================

download_municipality_boundaries()
```

```python
# ============================================================================
# Get dataset
# ============================================================================

dataset = get_dataset(
    "municipality_boundaries_raw"
)

print(dataset.local_path)
```

```python
# ============================================================================
# Load raw layers
# ============================================================================

communes_raw = gpd.read_file(
    dataset.local_path,
    layer="commune",
)

arrondissements_raw = gpd.read_file(
    dataset.local_path,
    layer="arrondissement_municipal",
)
```

```python
# ============================================================================
# Explore raw municipality layer
# ============================================================================

explore_dataframe(
    communes_raw,
)
```

```python
# ============================================================================
# Explore raw municipal arrondissement layer
# ============================================================================

explore_dataframe(
    arrondissements_raw,
)
```

```python
arrondissements_raw[
    arrondissements_raw["code_insee"].isin(
        [
            "69382",
            "75108",
            "13212",
        ]
    )
]
```

<span style="color:green">
<b>NOTE</b>:</br>
Saint-Barthélemy (97701) and Saint-Martin (97801) are overseas collectivities (COM). They legitimately have no EPCI, canton or arrondissement.
</span>

```python
# ============================================================================
# Prepare dataset
# ============================================================================

municipality_boundaries = prepare_municipality_boundaries(
    communes_raw,
    arrondissements_raw,
)
```

```python
# ============================================================================
# Explore prepared dataset
# ============================================================================

explore_dataframe(
    municipality_boundaries,
)
```

```python
municipality_boundaries[
    municipality_boundaries["insee_code"].isin(
        [
            "69123",
            "69382",
            "75056",
            "75108",
            "13055",
            "13212",
        ]
    )
][
    [
        "insee_code",
        "municipality",
        "entity_type",
        "parent_insee_code",
    ]
]
```

```python
# ============================================================================
# Validate prepared dataset
# ============================================================================

validate_dataframe(
    municipality_boundaries,
    key="insee_code",
    expect_missing=True,
    expected_missing_columns=[
        "parent_insee_code",
    ],
)
```

```python
# ============================================================================
# Save dataset
# ============================================================================

save_dataframe(
    municipality_boundaries,
    PREPARED_DIR / "municipality_boundaries.parquet",
)
```
