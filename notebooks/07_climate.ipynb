# Climate

This notebook prepares the monthly climate dataset used by HERMES.

## Pipeline

1. Download dataset
2. Load dataset
3. Explore raw dataset
4. Validate raw dataset
5. Prepare dataset
6. Explore prepared dataset
7. Validate prepared dataset
8. Save dataset

## Output

- `climate.parquet`

```python
# ============================================================================
# Imports
# ============================================================================

from hermes.config import PREPARED_DIR
from hermes.io import save_dataframe
from hermes.loaders import load_climate_raw
from hermes.preprocessing import prepare_climate
from hermes.sources import download_climate
from hermes.validation import (
    explore_dataframe,
    validate_dataframe,
)
```

```python
# ============================================================================
# Download dataset
# ============================================================================

download_climate()
```

```python
# ============================================================================
# Load dataset
# ============================================================================

climate_raw = load_climate_raw()
```

```python
# ============================================================================
# Explore raw dataset
# ============================================================================

explore_dataframe(
    climate_raw,
    title="Raw Climate Dataset",
)
```

```python
(
    climate_raw
    .isna()
    .sum()
    .loc[lambda s: s > 0]
    .sort_values(ascending=False)
)
```

```python
climate_raw[
    climate_raw.isna().any(axis=1)
].head()
```

```python
# ============================================================================
# Validate raw dataset
# ============================================================================

validate_dataframe(
    climate_raw,
    expect_missing=True,
)
```

<span style="color:green">
<b>Note on missing values:</b>

The SPI (Standardized Precipitation Index) and SSWI (Standardized Soil Wetness Index) variables are rolling climate indicators computed over several months.

At the beginning of each time series, these indicators are mathematically undefined because insufficient historical observations are available.

Therefore, missing values are expected for:

- SPI1
- SPI3
- SPI6
- SPI12
- SSWI1
- SSWI3
- SSWI6
- SSWI12

These missing values originate from the official Météo-France dataset and are not introduced by the HERMES preprocessing pipeline.
</span>

```python
# ============================================================================
# Prepare dataset
# ============================================================================

climate = prepare_climate(
    climate_raw,
)
```

```python
# ============================================================================
# Explore prepared dataset
# ============================================================================

explore_dataframe(
    climate,
    title="Prepared Climate Dataset",
)
```

```python
climate.columns.tolist()
```

```python
(
    climate
    .isna()
    .sum()
    .loc[lambda s: s > 0]
    .sort_values(ascending=False)
)
```

```python
climate.isna().sum().sum()
```

```python
climate.isna().mean().sort_values(ascending=False)
```

```python
# ============================================================================
# Validate prepared dataset
# ============================================================================

validate_dataframe(
    climate,
    key=[
        "grid_cell",
        "date",
    ],
    expect_missing=True,
)
```

```python
# ============================================================================
# Save dataset
# ============================================================================

save_dataframe(
    climate,
    PREPARED_DIR / "climate.parquet",
)
```
