# Population

This notebook prepares the municipality-level population dataset used by HERMES.

Pipeline

1. Download
2. Load
3. Explore
4. Validate raw dataset
5. Prepare
6. Validate prepared dataset
7. Save

Output

- `population.parquet`

```python
# ============================================================================
# Imports
# ============================================================================

from hermes.config import PREPARED_DIR
from hermes.io import save_dataframe
from hermes.loaders import load_population_raw
from hermes.preprocessing import prepare_population
from hermes.sources import download_population
from hermes.validation import (explore_dataframe, validate_dataframe)
```

```python
# ============================================================================
# Download dataset
# ============================================================================

download_population()
```

```python
# ============================================================================
# Load dataset
# ============================================================================

population_raw = load_population_raw()
```

```python
# ============================================================================
# Explore raw dataset
# ============================================================================

explore_dataframe(
    population_raw,
    title="Raw Population Dataset",
)
```

```python
# ============================================================================
# Validate raw dataset
# ============================================================================

validate_dataframe(
    population_raw,
)
```

```python
# ============================================================================
# Prepare dataset
# ============================================================================

population = prepare_population(
    population_raw,
)

population.head()
```

```python
# ============================================================================
# Explore prepared dataset
# ============================================================================

explore_dataframe(
    population,
    title="Prepared Population Dataset",
)
```

```python
# ============================================================================
# Validate prepared dataset
# ============================================================================

validate_dataframe(
    population,
    key="insee_code",
)
```

```python
# ============================================================================
# Save dataset
# ============================================================================

save_dataframe(
    population,
    PREPARED_DIR / "population.parquet",
)
```
