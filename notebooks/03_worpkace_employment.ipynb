# Workplace Employment

This notebook prepares the municipality-level workplace employment dataset used by HERMES.

## Pipeline

1. Download dataset
2. Load dataset
3. Explore raw dataset
4. Validate raw dataset
5. Prepare dataset
6. Explore prepared dataset
7. Validate prepared dataset
8. Save dataset

## Output

- `workplace_employment.parquet`

```python
# ============================================================================
# Imports
# ============================================================================

from hermes.config import PREPARED_DIR
from hermes.io import save_dataframe
from hermes.loaders import load_workplace_employment_raw
from hermes.preprocessing import prepare_workplace_employment
from hermes.sources import download_workplace_employment
from hermes.validation import (
    explore_dataframe,
    validate_dataframe,
)
```

```python
# ============================================================================
# Download dataset
# ============================================================================

download_workplace_employment()
```

```python
# ============================================================================
# Load dataset
# ============================================================================

workplace_employment_raw = load_workplace_employment_raw()
```

```python
# ============================================================================
# Explore raw dataset
# ============================================================================

explore_dataframe(
    workplace_employment_raw,
    title="Raw Workplace Employment Dataset",
)
```

```python
# ============================================================================
# Validate raw dataset
# ============================================================================

validate_dataframe(
    workplace_employment_raw,
    expect_missing=True,
)
```

```python
# ============================================================================
# Prepare dataset
# ============================================================================

workplace_employment = prepare_workplace_employment(
    workplace_employment_raw,
)
```

```python
# ============================================================================
# Explore prepared dataset
# ============================================================================

explore_dataframe(
    workplace_employment,
    title="Prepared Workplace Employment Dataset",
)
```

```python
# ============================================================================
# Validate prepared dataset
# ============================================================================

validate_dataframe(
    workplace_employment,
    key="insee_code",
)
```

```python
# ============================================================================
# Save dataset
# ============================================================================

save_dataframe(
    workplace_employment,
    PREPARED_DIR / "workplace_employment.parquet",
)
```
